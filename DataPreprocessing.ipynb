```python
# lets find out how many datas are missing per column
missing_count = data.isnull().sum()
print(missing_count)
```

```python
# We can see that 7 values are missing in cases
# When working with huge datasets it may be helpful to see what
# Percentage of the data is missing

total_data=np.prod(data.shape)
total_missing=missing_count.sum()
per_of_missing_data=(total_missing/total_data)* 100
print(per_of_missing_data)

# Conclusion: We can see we have a very small percentage 
# of missing data, we may delete those rows or fill them up
```
